## Corpus shape and class balance

Counts per split and class establish the project's scope and the **baseline to beat**: the accuracy a model gets by always predicting the most common class. IMDB is constructed 50/50, so if the splits preserved that balance the floor is 50% — which is what makes plain accuracy a usable headline metric for this comparison (with precision/recall/F1/ROC-AUC computed alongside in notebook 04).

```python
counts = (
    splits.groupby(["split", "label"])
    .size()
    .unstack()
    .rename(columns=CLASS_NAMES)
    .loc[["fit", "val", "test"]]
)
counts["total"] = counts.sum(axis=1)
counts.to_csv(TBL / "01-eda_split_balance.csv")

majority = counts[["negative", "positive"]].max(axis=1) / counts["total"]
print("Majority-class baseline accuracy per split:")
print(majority.map("{:.1%}".format).to_string())
counts
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
counts[["negative", "positive"]].plot.bar(
    ax=ax, color=[COLORS["negative"], COLORS["positive"]]
)
ax.set_ylabel("reviews")
ax.set_xlabel("split")
ax.set_title("Class balance by split")
ax.tick_params(axis="x", rotation=0)
fig.tight_layout()
fig.savefig(FIG / "01-eda_split_balance.png")
```

## Review length by class

Words per review (simple whitespace tokens — interpretable and tokenizer-free). Two things matter downstream: the **long tail** motivates capping the TF-IDF vocabulary (`max_features ≈ 20k`) rather than keeping every rare term, and any length *difference between classes* would be a shortcut feature the models could lean on. From here on, all content analysis uses the fit split only.

```python
fit = splits[splits["split"] == "fit"].copy()
fit["clean"] = fit["text"].map(preprocess_text)

# Length is measured on the cleaned text, because that is what the models see.
# Raw counts are kept only to quantify how much the markup inflates them: <br />
# tags sit glued to neighbouring words, so a doubled tag costs one extra token
# on a whitespace split, not four.
fit["n_words"] = fit["clean"].str.split().str.len()
inflation = fit["text"].str.split().str.len() - fit["n_words"]

length_summary = (
    fit.groupby(fit["label"].map(CLASS_NAMES))["n_words"]
    .describe(percentiles=[0.25, 0.5, 0.75, 0.95, 0.99])
    .round(1)
    .rename_axis("class")
)
length_summary.to_csv(TBL / "01-eda_length_summary.csv")

print(
    f"markup inflation on raw counts: {inflation.sum():,} tokens total, "
    f"{inflation.mean():.1f}/review; {(inflation > 0).sum():,} reviews affected "
    f"(median {inflation[inflation > 0].median():.0f} tokens among those)"
)
length_summary
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2), layout="constrained")

# Left: the bulk, where the class-overlap story lives. Reviews past 800 words
# fall outside the bin range rather than being clipped into it -- clipping piles
# them onto the boundary and reads as a spike that is not there.
for label, name in CLASS_NAMES.items():
    axes[0].hist(
        fit.loc[fit["label"] == label, "n_words"],
        bins=np.arange(0, 820, 20), alpha=0.55, color=COLORS[name], label=name,
    )
axes[0].set_xlabel("words per review (view truncated at 800)")
axes[0].set_ylabel("reviews")
axes[0].set_title("Bulk: the two classes overlap almost completely")
axes[0].legend()

# Right: full range on a log count axis. Tail bins hold one or two reviews each,
# invisible on a linear axis, so the log scale is what makes them legible. The
# final edge is derived from the data so the "nothing omitted" claim cannot
# silently become false if the corpus ever changes.
tail_bins = np.arange(0, fit["n_words"].max() + 26, 25)
assert fit["n_words"].max() <= tail_bins[-1], "tail bins must cover the longest review"
for label, name in CLASS_NAMES.items():
    axes[1].hist(
        fit.loc[fit["label"] == label, "n_words"],
        bins=tail_bins, alpha=0.55, color=COLORS[name], label=name,
    )
axes[1].set_yscale("log")
axes[1].set_xlabel("words per review (full range, nothing omitted)")
axes[1].set_ylabel("reviews (log)")
axes[1].set_title("Tail: 25-word bins on a log count axis")
axes[1].legend()

fig.suptitle("Review length by class (fit split)")
fig.savefig(FIG / "01-eda_length_by_class.png")
```

## Markup artifacts

IMDB reviews were scraped from the web, and line breaks survive as literal `<br />` tags. This is the classic annotation-artifact problem: untreated, the tokenizer happily turns markup into a *feature*. First the prevalence sweep, then the demonstration — the top-20 unigrams with and without stripping — which is the empirical justification for the `strip <br />` rule frozen in `docs/decisions.md`.

```python
patterns = {
    "<br /> line break": r"<br\s*/?>",
    "any <tag>": r"<[a-zA-Z][^>]*>",
    "HTML entity (&...;)": r"&(?:#\d+|[a-zA-Z]+);",
    "non-ASCII characters": r"[^\x00-\x7F]",
}
counts = {
    name: int(fit["text"].str.contains(pat, regex=True).sum())
    for name, pat in patterns.items()
}
# The "any <tag>" row overlaps the <br /> row almost entirely, so differencing the
# two proves nothing. This second pass answers the disjoint question instead:
# what markup is still standing once line breaks are stripped?
no_br = fit["text"].str.replace(r"<br\s*/?>", " ", regex=True)
counts["<tag> surviving <br /> removal"] = int(
    no_br.str.contains(r"<[a-zA-Z][^>]*>", regex=True).sum()
)

artifacts = pd.DataFrame({"reviews_containing": counts})
artifacts["pct_of_fit"] = (artifacts["reviews_containing"] / len(fit) * 100).round(1)
artifacts.to_csv(TBL / "01-eda_html_artifacts.csv", index_label="artifact")
artifacts
```

```python
def top_unigrams(texts, k=20):
    cv = CountVectorizer(lowercase=True)
    x = cv.fit_transform(texts)
    totals = np.asarray(x.sum(axis=0)).ravel()
    vocab = np.array(cv.get_feature_names_out())
    order = np.argsort(totals)[::-1][:k]
    return pd.Series(totals[order], index=vocab[order])


raw_top = top_unigrams(fit["text"])
stripped_top = top_unigrams(fit["text"].map(preprocess_text))

fig, axes = plt.subplots(1, 2, figsize=(9, 5.5), sharex=True, layout="constrained")
panels = [
    (axes[0], raw_top, "raw text"),
    (axes[1], stripped_top, "after preprocess_text"),
]
for ax, top, title in panels:
    colors = ["indianred" if w == "br" else "grey" for w in top.index]
    ax.barh(top.index[::-1], top.values[::-1], color=colors[::-1])
    ax.set_title(title)
    ax.set_xlabel("total occurrences (fit split)")
    ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v / 1000:.0f}k"))
fig.suptitle("Top-20 unigrams before vs. after shared preprocessing")
fig.savefig(FIG / "01-eda_br_artifact_demo.png")

rank = int(np.where(raw_top.index == "br")[0][0]) + 1 if "br" in raw_top.index else None
print(f"'br' rank among raw unigrams: {rank} ({raw_top.get('br', 0):,} occurrences)")
```

## Top n-grams by class

Raw counts (`CountVectorizer`, not the TF-IDF artifact — this notebook reads no feature artifacts) on stripped, lowercased text with stop words kept, matching the shared preprocessing choices. Two views:

1. **Most frequent** unigrams and bigrams per class — the required item. Expect heavy overlap: both classes talk about films in the same function words.
2. **Most distinctive** n-grams per class via a smoothed log-ratio of class frequencies — this is the view that previews the sentiment lexicon, i.e., the words logistic regression's coefficients should later recover in notebook 02.

```python
# fit["clean"] was computed with the length summary above.
NGRAM_TYPE = {1: "unigram", 2: "bigram"}


def top_ngrams_by_class(n, k=20):
    cv = CountVectorizer(lowercase=True, ngram_range=(n, n), min_df=2)
    x = cv.fit_transform(fit["clean"])
    vocab = np.array(cv.get_feature_names_out())
    rows = []
    for label, name in CLASS_NAMES.items():
        mask = (fit["label"] == label).to_numpy()
        totals = np.asarray(x[mask].sum(axis=0)).ravel()
        order = np.argsort(totals)[::-1][:k]
        rows += [
            {
                "class": name,
                "ngram_type": NGRAM_TYPE[n],
                "rank": r + 1,
                "ngram": vocab[i],
                "count": int(totals[i]),
            }
            for r, i in enumerate(order)
        ]
    return pd.DataFrame(rows)


top_ngrams = pd.concat(
    [top_ngrams_by_class(1), top_ngrams_by_class(2)], ignore_index=True
)
top_ngrams.to_csv(TBL / "01-eda_top_ngrams_by_class.csv", index=False)

preview = top_ngrams[top_ngrams["rank"] <= 10].assign(
    item=lambda d: d["ngram"] + "  (" + d["count"].astype(str) + ")"
)
preview.pivot(index="rank", columns=["ngram_type", "class"], values="item")
```

```python
cv = CountVectorizer(lowercase=True, ngram_range=(1, 2), min_df=2)
x = cv.fit_transform(fit["clean"])
vocab = np.array(cv.get_feature_names_out())
print(f"vocabulary size (uni+bigrams, min_df=2): {len(vocab):,} types")
neg_mask = (fit["label"] == 0).to_numpy()
neg = np.asarray(x[neg_mask].sum(axis=0)).ravel()
pos = np.asarray(x[~neg_mask].sum(axis=0)).ravel()

# Smoothed log-ratio of relative class frequencies (add-1). A frequency floor
# keeps rare n-grams from dominating the ranking on unstable ratios.
log_ratio = np.log((pos + 1) / (pos.sum() + len(vocab))) - np.log(
    (neg + 1) / (neg.sum() + len(vocab))
)
frequent = np.where(neg + pos >= 50)[0]
order = frequent[np.argsort(log_ratio[frequent])]
sel = np.concatenate([order[:15], order[-15:]])

fig, ax = plt.subplots(figsize=(7, 7))
vals = log_ratio[sel]
colors = [COLORS["negative"] if v < 0 else COLORS["positive"] for v in vals]
ax.barh(vocab[sel], vals, color=colors)
ax.axvline(0, color="black", lw=0.8)
ax.set_xlabel("log-ratio of class frequency  (\u2190 negative | positive \u2192)")
ax.set_title("Most class-distinctive n-grams (fit split, \u226550 occurrences)")
fig.tight_layout()
fig.savefig(FIG / "01-eda_distinctive_ngrams.png")
```

## Data-quality notes

Report-only checks on the fit split: exact-duplicate review texts and near-empty reviews. No action is taken here — the split spec is frozen in `docs/decisions.md`, so anything nonzero becomes a documented limitation rather than a silent data mutation.

```python
dup_counts = fit["text"].value_counts()
dup_texts = dup_counts[dup_counts > 1]
dup_rows = int(dup_texts.sum() - len(dup_texts))
conflicting = int(
    fit[fit["text"].isin(dup_texts.index)]
    .groupby("text")["label"]
    .nunique()
    .gt(1)
    .sum()
)
near_empty = int((fit["n_words"] < 5).sum())

print(f"within fit -- exact-duplicate texts: {len(dup_texts)} appear more than once")
print(f"  -> {dup_rows} redundant rows; {conflicting} duplicated with BOTH labels")
print(f"reviews under 5 words: {near_empty}")

# Corpus integrity across all three splits. Grouped on normalised text and never
# printed, so no test review is read here. Duplicates straddling fit/val and test
# are a property of the published benchmark rather than of our sampling, and are
# reported rather than repaired: the split spec is frozen, and trimming it would
# break comparability with the standard IMDB 25k test set.
normalised = splits["text"].map(preprocess_text)
spans = splits.groupby(normalised)["split"].agg(lambda s: tuple(sorted(set(s))))
cross_split = spans[spans.map(len) > 1].value_counts()
label_conflicts = int(splits.groupby(normalised)["label"].nunique().gt(1).sum())

print("\nacross splits -- texts appearing in more than one split:")
for pair, n in cross_split.items():
    print(f"  {' + '.join(pair):14s} {n}")
print(f"  conflicting labels among all duplicates: {label_conflicts}")
print(f"  -> at most {cross_split.get(('fit', 'test'), 0) / 25_000:.2%} of the test "
      "split is memorisable from fit")

# Is length itself a usable class signal? ROC-AUC of length used directly as a
# ranking score. This is a rank statistic, so nothing is fitted here.
print(
    f"\nlength-only ROC-AUC (fit): {roc_auc_score(fit['label'], fit['n_words']):.4f}"
    "  (0.5 = no signal)"
)
```

## Key takeaways

- **Balance:** every split is exactly 50/50, so the majority-class floor is 50% and plain accuracy is a legitimate headline metric for the model comparison (the full metric suite still lands in notebook 04).
- **Length is not a shortcut feature:** measured on the cleaned text the models actually see, the class distributions overlap almost completely (median 174 words negative vs. 170 positive). Used alone as a ranking score, length scores **ROC-AUC 0.491** on the fit split — indistinguishable from chance, so the claim rests on a statistic rather than on how the histograms look. The right tail is long and slightly heavier for positive reviews — 95th percentile 569 vs. 611, 99th 877 vs. 905, max 1,347 vs. 2,459 — which is part of why the shared TF-IDF vocabulary is capped at ~20k features rather than keeping every rare term. Measuring on raw text instead inflates counts by only 2.3 tokens per review (22,516 in total across the 5,816 reviews carrying markup), moving each median by at most two words: `<br />` tags sit glued to neighbouring words, so a doubled tag costs one extra whitespace token, not four.
- **Markup:** 58% of fit reviews contain `<br />`, and untreated it lands as the **#6 unigram overall (~40k occurrences)** — the annotation-artifact problem in one bar chart. Stripping it is the entire markup job: **once `<br />` is removed, no fit review contains a real HTML tag.** The two hits that survive the naive `<tag>` regex are false positives, a bare URL in angle brackets and a stray `(sp?)` typo, and HTML entities appear in exactly two reviews. What remains is character-level rather than markup: 9.5% of reviews carry non-ASCII bytes, mostly `é` plus cp1252 mojibake standing in for en-dashes and ellipses. Accented letters survive into tokens; the mojibake bytes act as separators and fragment the words they sit inside. Left untreated, because the preprocessing spec is frozen in `docs/decisions.md` — a documented limitation, not a silent fix.
- **Lexicon preview, and entity leakage:** the class-distinctive n-grams are of two kinds. Genuine sentiment vocabulary ("loved this", "highly recommended", "well worth" vs. "waste of", "avoid this", "worst movie") previews what LR's coefficients should recover in notebook 02. But proper nouns rank just as high: Powell, Gandhi, Spock, and Matthau on the positive side, while "boll" (the director Uwe Boll) is the single most negative-distinctive term in the corpus. Bag-of-words models will learn *entity priors* alongside sentiment — a lens for reading LR's coefficients, a candidate explanation for model disagreements (notebook 05's `other` tag), and a real limitation for the recommender framing, since entity priors do not transfer beyond this corpus.
- **Data quality, and a benchmark caveat:** within the fit split, 17 duplicated review texts (18 redundant rows, none with conflicting labels) and zero near-empty reviews. Across splits, the published benchmark duplicates *itself*: 48 texts appear in both `fit` and `test`, 32 in both `val` and `test`, and 13 in both `fit` and `val`, with no label conflicts anywhere. So at most **0.19% of the test split is memorisable from training data** — immaterial to the headline comparison, but reported rather than repaired: the split spec is frozen, and trimming duplicates would break comparability with the standard IMDB 25k test set.
